```python
## A little more complex example - part 1: 
# getting all links for images on the reptil database
site_reptil = 'http://www.reptile-database.org/'
req = requests.get(site_reptil)
list_images = re.findall(r'[:\w\-.//]+jpg', req.text)
list_images
```

```python
## a little more complex example, part 2:
## saving all image links as images on pc

import matplotlib.pyplot as plt
import matplotlib.image as mpimg

for image in list_images:
    req = requests.get(image)
    with open(image.split(r'/')[-1], 'wb') as x:
        x.write(req.content)
    
    img = mpimg.imread(image.split(r'/')[-1])
    imgplot = plt.imshow(img)
    plt.show()
```
