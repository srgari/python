```python
import pandas as pd 
import numpy as np 

df = pd.read_csv('venda_produtos.csv')
df
```

```python
df2 = df.pivot_table(index = 'cpf', columns = 'produto', values = 'preco', aggfunc = 'count').reset_index().notna()
df2
```

```python
df3 = pd.crosstab(
    index = df2['maçã'],
    columns = df2['banana']
)
df4 = df3/388
display(df3, df4)
```

```python
df4.loc[True].sum(), df4[True].sum(), df3.loc[True].sum(), df3[True].sum()
```
