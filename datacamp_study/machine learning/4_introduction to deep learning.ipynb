Neural network interactions

Forward propagation
- input, hidden layer, output 
- each input interacts with each hidden layer
- each hidden layer interact with the output

I = Input  
HL = Hidden Layer

Only input has values
Each interaction input - hidden layer has a weight  
Ex: Input 1 has interaction of 1 with Hidden layer 1

process = sum(input values * weight of each interaction)

I1-HL1  
....X.......>.Output   
I2-HL2  

I1 = 2  
I2 = 3
W(I1 - HL1) = 1  
W(I2 - HL1) = 1  
W(I1 - HL2) = -1  
W(I2 - HL2) = 1  

I1 - HL1  
I1 - HL2  
I2 - HL1  
I2 - HL2  

I1(2) -(+1)- HL1   (2*1 =  2)  
I2(3) -(+1)- HL1   (3*1 =  3)  
I1(2) -(-1)- HL2   (2*-1= -2)  
I2(3) -(+1)- HL2   (3*1 =  3)  

HL1 = +2+3 = 5  
HL2 = -2+3 = 1  

HL1(5) -(+2)- Output
HL1(1) -(-1)- Output

Output = +10 - 1 = 9

````python
# activattion function: relu
def relu(input): return max(input, 0)

def predict_with_network(input_data_row, weights):
    # calculate node 0 value: node_0_output
    node_0_input = input_data_row @ weights['node_0']
    node_0_output = relu(node_0_input)
    # calculate node 1 value: node_1_output
    node_1_input = input_data_row @ weights['node_1']
    node_1_output = relu(node_1_input)

    # put node values into array
    hidden_layer_outputs = np.array([node_0_output, node_1_output])

    # calculate model output (do not apply relu)
    input_to_final_layer = hidden_layer_outputs @ weights['output']
    model_output = relu(input_to_final_layer)
    return model_output

input_data = [
    np.array([3, 5]), 
    np.array([ 1, -1]), 
    np.array([0, 0]), 
    np.array([8, 4])]

weights =  {
    'node_0': np.array([2, 4]), 
    'node_1': np.array([ 4, -5]), 
    'output': np.array([2, 7])
    }
results = []
for input_data_row in input_data:
    results.append(predict_with_network(input_data_row, weights))
results       
````

````python
inputs = np.array([1,1])
weights = {
    'node0_0': np.array([2,4]),
    'node0_1': np.array([4, -5]),
    'node1_0': np.array([0,1]),
    'node1_1': np.array([1,1]),
    'output': np.array([5,1])
}

node0_0 = inputs @ weights['node0_0']
node0_1 = inputs @ weights['node0_1']
hidden_layer_0 = np.array([node0_0, node0_1])
node1_0 = hidden_layer_0 @ weights['node1_0']
node1_1 = hidden_layer_0 @ weights['node1_1']
hidden_layer_1 = np.array([node1_0, node1_1])
output = hidden_layer_1 @ weights['output']
print(f'{node0_0 = }')
print(f'{node0_1 = }')
print(f'{hidden_layer_0 = }')
print(f'{node1_0 = }')
print(f'{node1_1 = }')
print(f'{hidden_layer_1 = }')
print(f'{output = }')
````

````python
def predict_with_network(input_data, weights):
    # Calculate node 0 in the first hidden layer
    node_0_0_input = (input_data * weights['node_0_0']).sum()
    node_0_0_output = relu(node_0_0_input)

    # Calculate node 1 in the first hidden layer
    node_0_1_input = input_data @ weights['node_0_1']
    node_0_1_output = relu(node_0_1_input)

    # Put node values into array: hidden_0_outputs
    hidden_0_outputs = np.array([node_0_0_output, node_0_1_output])
    
    # Calculate node 0 in the second hidden layer
    node_1_0_input = hidden_0_outputs @ weights['node_1_0']
    node_1_0_output = relu(node_1_0_input)

    # Calculate node 1 in the second hidden layer
    node_1_1_input = hidden_0_outputs @ weights['node_1_1']
    node_1_1_output = relu(node_1_1_input)

    # Put node values into array: hidden_1_outputs
    hidden_1_outputs = np.array([node_1_0_output, node_1_1_output])

    # Calculate model output: model_output
    model_output = hidden_1_outputs @ weights['output']
    
    # Return model_output
    return(model_output)

input_data = np.array([3,5])
weights = {
    'node_0_0': np.array([2, 4]),
    'node_0_1': np.array([ 4, -5]),
    'node_1_0': np.array([-1,  2]),
    'node_1_1': np.array([1, 2]),
    'output':   np.array([2, 7])}
output = predict_with_network(input_data, weights)
print(output)
````

````python
# activattion function: relu
def relu(input): return max(input, 0)

def predict_with_network(input_data_row, weights):
    # calculate node 0 value: node_0_output
    node_0_input = input_data_row @ weights['node_0']
    node_0_output = relu(node_0_input)
    # calculate node 1 value: node_1_output
    node_1_input = input_data_row @ weights['node_1']
    node_1_output = relu(node_1_input)

    # put node values into array
    hidden_layer_outputs = np.array([node_0_output, node_1_output])

    # calculate model output (do not apply relu)
    input_to_final_layer = hidden_layer_outputs @ weights['output']
    model_output = relu(input_to_final_layer)
    return model_output

# the data point you will make a prediction for
input_data = np.array([0,3])

# sample weights
weights_0 = {
    'node_0':[2,1],
    'node_1':[1,2],
    'output':[1,1],
}

# the actual target value, used to calculate the error
target_actual = 3

# make prediction using original weights
model_output_0 = predict_with_network(input_data, weights_0)

# calculate error: error_0
error_0 = model_output_0 - target_actual

# create weights that cause the network to make perfect prediction (3): weights_1
weights_1 = {
    'node_0':[2,1],
    'node_1':[1,0],
    'output':[1,1],
}

# make prediction using new weights: model_output_1
model_output_1 = predict_with_network(input_data, weights_1)

# calculate error: error_1
error_1 = target_actual - model_output_1

# print error_0 and error_1
print(error_0, error_1)
````

````python
# activattion function: relu
import numpy as np 
from sklearn.metrics import mean_squared_error
def relu(input): return max(input, 0)

def predict_with_network(input_data_row, weights):
    # calculate node 0 value: node_0_output
    node_0_input = input_data_row @ weights['node_0']
    node_0_output = relu(node_0_input)
    # calculate node 1 value: node_1_output
    node_1_input = input_data_row @ weights['node_1']
    node_1_output = relu(node_1_input)

    # put node values into array
    hidden_layer_outputs = np.array([node_0_output, node_1_output])

    # calculate model output (do not apply relu)
    input_to_final_layer = hidden_layer_outputs @ weights['output']
    model_output = relu(input_to_final_layer)
    return model_output

target_actuals = [1, 3, 5, 7]
input_data = [np.array([0, 3]), np.array([1, 2]), np.array([-1, -2]), np.array([4, 0])]

weights_0 = {'node_0': np.array([2, 1]), 'node_1': np.array([1, 2]), 'output': np.array([1, 1])}
weights_1 = {'node_0': np.array([2, 1]), 'node_1': np.array([1, 1.5]), 'output': np.array([1, 1.5])}


model_output_0 = []
model_output_1 = []
for row in input_data:
    predict_0 = predict_with_network(row, weights_0)
    model_output_0.append(predict_0)
    predict_1 = predict_with_network(row, weights_1)
    model_output_1.append(predict_1)

mse_0 = mean_squared_error(target_actuals, model_output_0)
mse_1 = mean_squared_error(target_actuals, model_output_1)

print(mse_0, mse_1)
````

Gradient descent

Slope calculation example: 
- node: 3
- weight: 2
- output: 6
- actual target: 10

- to calculate the slope for a weight, you need to multiply:
    - slope of the loss function with respect to (wtr) value at the node we feed into
    - the value of the node that feeds into our weight
    - slope of the activation function wrt value we feed into

Gradient descent: 
- derivative of MSE; intercept = 0
- check your Joker notebook
- gradient = slope
- Mean squared error: 
    - MSE(weight) = (actual-predicted)**2
    - MSE(weight) = ((node * weight)-predicted)**2
    - dMSE/d(weight) = - 2*( (node * weight))*node 
    - dMSE/dx = - 2*((3*2) - 10 ) * 3 = - 2*(10-6) * 3 = -24
