# chapter 3

- Coefficient of determination, or R squared, or R2
  - equals to correlation squared!
  - r squared: 1 explanatory variable
  - R squared: 2 or explanatory variables

- Residual standard error (RSE)
  - standard error, but calculated using residual instead of mean
  - 'typical' difference between prediction and observed response
  

- Mean Squared Error (MSE):
  - squared residual standard error
  - residual variance
  - variance of the residual
  - squared standard deviation of residual
  - mean of RSE squared

- Residual standard error, or MSE (mean squared error)
  - squared root of mean squared error
  
- Python code for RSE and MSE:
```python
mse = mdl_bream.mse_resid # mean squared error
rse = np.sqrt(mse) # root mean squared error
```

- degrees of freedom: len of data minus 2
  - number of observations minus number of model coefficients
dicted and observed bream masses is tipically about 74g

- Python code for R squared:
```python
from statsmodels.formula.api import ols
mdl_bream = ols('mass_g ~ length_cm', data = bream).fit()
mdl_bream.summary() # summary
mdl_bream.rsquared # rsquared
bream['length_cm'].corr(bream['mass_g']) ** 2 ## rsquared
```


- RSS:
  - Residual Squared sum
  - sum(mdl_bream.resid**2)

- Degrees of freedom: 
  - len(data) - model coeffs
  - len(bream) - 2

- RSE:
  - Residual Standard Error 
  - MSE = RSE**2
    - RSE = MSE ** 0.5
    - RSE = (RSS/deg_freedom)**0.5

- RMSE
  - Root-mean-square error  
  - RMSE: RSS/n_obs


- Python code for MSE RSE:
```python
mse = mdl_bream.mse_resid
rse = np.sqrt(mse)
mean_resid_sum_of_sq = sum(mdl_bream.resid ** 2)/len(mdl_bream)
```
- python code for RSE using degrees of freedom
```python
residuals_sq = mdl_bream.resid ** 2
resid_sum_of_sq = sum(residuals_sq)
deg_freedom = len(bream.index) - 2
rse = np.sqrt(resid_sum_of_sq/deg_freedom)
```

- Odds ratios:
  - probability of something happening divided by the probability of not happening
  - odds_ratio = probability/(1-probability)

```python
prediction_data['odds_ratio'] = prediction_data['has_churned']/(1-prediction_data['has_churned'])

prediction_data['log_odds_ratio'] = np.log(prediction_data['odds_ratio'])
```

- sensitivity = (TP) / (FN + TP)
- specificity = (TN) / (TN + FP)
