Template

```python
from string import Template
a1, a2, a3 = 'lala lele lili'.split()
price = 20
sentence = Template('my favorite gals are $gal1, ${gal2} and ${gal3}_doll!\nThey only cost $$ ${price}')

final_sentence = sentence.substitute(
    gal1=a1,
    gal2=a2,
    gal3=a3,
    price=price
)
print(final_sentence)
```

Regex: greedy vs non-greedy

```python
import re 
tags = '<strong> lala </strong>'
greedy_tag = re.findall('<(.+)>',tags)
nongreedy_tag = re.findall('<(.+?)>',tags)
greedy_tag, nongreedy_tag
```

```python
sentence = 'I am here. You are here. We are here.'
greedy_sentence = re.findall('\.(.+)\.', sentence)
nongreedy_sentence = re.findall('\.(.*?)\.',sentence)
greedy_sentence, nongreedy_sentence
```

Backreferences

```python
date = 'when was 12-03-2008?'
re.search('(\d{2})-(\d{2})-(\d{4})', date).group(1)
```

```python
date = 'when was 12-03-2008?'
(re.search('(?P<day>\d{2})-(?P<month>\d{2})-(\d{4})', date).group('day'),
re.search('(?P<day>\d{2})-(?P<month>\d{2})-(\d{4})', date).group('month'))
```

```python
# Only works because 12- is identical to 12-
# useful to capture repeated words
date = 'when was 12-12-2008?'
re.findall(r'(\d{2})-\1.*', date)
```

Lookaround

```python
text = 'the white cat sat on the chair and the black cat sat on the floor'
#look-ahead
re.findall('\w+\s\w+\s(?=sat)',text)
```

```python
text = 'the white cat sat on the chair and the black cat sat on the floor'
#look-ahead
re.findall('(?<=cat sat)\s\w+\s\w+\s\w+',text)
```

```python
text = 'the white cat sat on the chair and the black cat sat on the floor'
#look-behind: two variables
re.findall('(?:(?<=white cat sat)|(?<=black cat sat))\s\w+\s\w+\s\w+', text)
```
