```python
# requires visual studio and C++ enabled
from wordcloud import WordCloud 
import matplotlib.pyplot as plt 
```

```python
import requests 
req = requests.get('https://www.folha.uol.com.br')
req.encoding = req.apparent_encoding
```

```python
import re
text = req.text#.encode('latin-1').decode('utf-16')

words = set(re.findall('<h[12].*>(.*)</h[12]>', text))
words
```

```python
from textblob import TextBlob
[(x, TextBlob(x).sentiment) for x in words]
```

```python
all_words = ' '.join(words)
img_words = WordCloud().generate(all_words)
plt.imshow(img_words, aspect = 'auto', interpolation = 'nearest')
plt.show()
```
