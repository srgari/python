```python
import pandas as pd 
import re 
import requests
```

```python
url = 'https://dragonballfighterz.fandom.com/wiki/Dramatic_Finish'
req = requests.get(url)
html = req.text
```

```python
fighters = re.findall('class="mw-headline" id=(?!"Conditions|"Description|"Script|"Reference)"(.*)"',html)
print(len(fighters))
fighters
```

```python
x = re.findall('Winning.*Character.*a href="/wiki/(.*?)"',html)
print(len(x))
x
```

```python
winning_character = re.findall('Winning.*Character.*a href="/wiki/(.*?)"',html)
print(len(winning_character))
winning_character
```

```python
losing_character = re.findall('Losing.*Character.*a href="/wiki/(.*?)"',html)
print(len(losing_character))
losing_character
```

```python
stage = re.findall('Stage.*a href="/wiki/(.*?)"',html)
sorted(stage)
```

```python
df = pd.DataFrame({
    'stage': stage,
    'fighters':fighters,
    'winning_character': winning_character,
    'losing_character': losing_character,
})
df.sort_values('stage')
df['stage'] = df['stage'].str.replace('%27',"'").str.replace('_'," ")
```

```python
df.to_csv('dragon_ball_fighterz_dramatic_finish.csv', index = False)
```
